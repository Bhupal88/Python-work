# CALCULATOR

```python
def add(x,y):
    return x + y

def subtract(x,y):
    return x - y

def multiply(x,y):
    return x * y

def divide(x,y):
    return x / y
```

```python
choice=input('Enter your choice (A, S, D, M): ')

num1=int(input('Enter the first number: '))

num2=int(input('Enter the second number: '))
```

```python
print('Enter A for Addition.')
print('Enter S for Subtraction.')
print('Enter M for Multiplication.')
print('Enter D for Division.')


while True:
    choice=input('Enter your choice (A, S, D, M): ')
    if choice.upper() in ('ADD','ADDITION','A','SUBTRACT','SUBTRACTION','S','DIVIDE','DIVISION','D','MULTIPLY','MULTIPLICATION','M'):
        num1=float(input('Enter the first number: '))
        num2=float(input('Enter the second number: '))

        if choice.upper() in ('ADD','ADDITION','A'):
            print('Result:' ,num1, '+', num2, '=' ,add(num1,num2))
        elif choice.upper() in ('SUBTRACT','SUBTRACTION','S'):
            print('Result:' ,num1, '-', num2, '=', subtract(num1,num2))
        elif choice.upper() in ('MUTLIPLY','MULTIPLICATION','M'):
            print('Result:' ,num1, '*', num2, '=', multiply(num1,num2))
        elif choice.upper() in ('DIVIDE','DIVISION','D'):
            print('Result:' ,num1, '/', num2, '=', divide(num1,num2))
    else:
        print('Please enter a valid choice')

    next_calc=input('Do you want to do another calculation? (yes/no)')
    if next_calc.lower() in ('no','n'):
        break
```
