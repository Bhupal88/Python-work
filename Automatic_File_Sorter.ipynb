```python
path = r'C:\Users\Bhupal\OneDrive\Desktop\Cafe_Website_Project\\'
```

```python
import os, shutil
```

```python
os.listdir(path)
```

```python
folder_names=['HTML Files','Image Files','CSS Files', 'Misc Files']

for folder in folder_names:
    if not os.path.exists(path + folder):
        os.makedirs(path + folder)
```

```python
file_names=os.listdir(path)
```

```python
for file in file_names:
    if '.html' in file and not os.path.exists(path + 'HTML Files\\' + file):
        shutil.move(path+file, path + 'HTML Files\\' + file)
    elif '.css' in file and not os.path.exists(path + 'CSS Files\\' + file):
        shutil.move(path+file, path + 'CSS Files\\' + file)
    elif ('.png' in file or '.jpg' in file or '.PNG' in file) and not os.path.exists(path + 'Image Files\\' + file):
        shutil.move(path+file, path + 'Image Files\\' + file)
    elif ('.wav' in file or '.py' in file) and not os.path.exists(path + 'Misc Files\\' + file):
        shutil.move(path+file, path + 'Misc Files\\' + file)
```
