# Automated Crypto Web Scraper

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
from datetime import datetime
import time
import os
```

```python

def automated_crypto_pull():
    url='https://coinmarketcap.com/currencies/bitcoin/'
    page=requests.get(url)
    soup=BeautifulSoup(page.text,'html')
    
    crypto_name=soup.find('span',class_='sc-65e7f566-0 lsTl').text
    crypto_price=soup.find('span',class_='sc-65e7f566-0 WXGwg base-text').text
    final_price=crypto_price.replace('$','')
    
    dt=datetime.now()
    
    dict={'Crypto Name': crypto_name,
          'Price': final_price,
         'TimeStamp': dt}
    df=pd.DataFrame([dict])
    df
    
    if os.path.exists(r'C:\Users\Bhupal\Python Tutorial\Crypto Web Puller\Crypto_Automated_Puller.csv'):
        df.to_csv(r'C:\Users\Bhupal\Python Tutorial\Crypto Web Puller\Crypto_Automated_Puller.csv', mode='a', header=False, index=False)
    else:
        df.to_csv(r'C:\Users\Bhupal\Python Tutorial\Crypto Web Puller\Crypto_Automated_Puller.csv', index=False)
    print(df)
```

```python
while True:
    automated_crypto_pull()
    time.sleep(10)
```
